```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import pickle as pkl
```

```python
df = pkl.load(open('project8.pkl', 'rb'))
df_aug = pkl.load(open('project8_augmented.pkl', 'rb'))
```

```python
#count plot for subreddit
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Plot for original dataframe
sns.countplot(ax=axes[0], x='subreddit', data=df)
axes[0].set_title('Number of posts per subreddit')

# Plot for augmented dataframe
sns.countplot(ax=axes[1], x='subreddit', data=df_aug)
axes[1].set_title('Number of posts per subreddit (augmented)')

plt.show()
```
